```python
import json
```

```python
groundFile = json.load(open("GroundTiles.json"))
surface = json.load(open("SurfaceTiles.json"))
surface2 = json.load(open("SurfaceTiles_v2.json"))
```

```python
newGround = []
newSurface = []
newSurface2 = []
```

```python
for row in range(50):
    newGround.append([])
    newSurface.append([])
    newSurface2.append([])
    for column in range(49, -1, -1):
        newGround[row].append(groundFile[row][column])
        newSurface[row].append(surface[row][column])
        newSurface2[row].append(surface2[row][column])

for row in range(50):
    for column in range(50):
        newGround[row][column]["column"] = column
        if newSurface[row][column] != 0:
            newSurface[row][column]["column"] = column
        if newSurface2[row][column] != 0:
            newSurface2[row][column]["column"] = column
        
```

```python
with open('../sources/GroundTiles.json', 'w') as outfile:
    json.dump(newGround, outfile)
with open('../sources/SurfaceTiles.json', 'w') as outfile:
    json.dump(newSurface, outfile)
with open('../sources/SurfaceTiles_v2.json', 'w') as outfile:
    json.dump(newSurface2, outfile)
```
